## Récupération des données

```julia
function readKnaptxtInstance(filename)
    price=Int64[]
    weight=Int64[]
    KnapCap=Int64[]
    open(filename) do f
        for i in 1:3
            tok = split(readline(f))
            if(tok[1] == "ListPrices=")
                for i in 2:(length(tok)-1)
                    push!(price,parse(Int64, tok[i]))
                end
            elseif(tok[1] == "ListWeights=")
                for i in 2:(length(tok)-1)
                    push!(weight,parse(Int64, tok[i]))
                end
            elseif(tok[1] == "Capacity=")
                push!(KnapCap, parse(Int64, tok[2]))
            else
                println("Unknown read :", tok)
            end
        end
    end
    capacity=KnapCap[1]
    return price, weight, capacity
end
```

## Code

```julia
function sacADosDynamique(capacite, poid, valeur)
    n=length(poid)
    SousProbleme = zeros(n+1,capacite+1)
    for elt in 1:n
        for c in 2:(capacite+1)
            if c > poid[elt]
                SousProbleme[elt+1, c] = max(SousProbleme[elt,c], SousProbleme[elt, (c-poid[elt])]+valeur[elt])
            else
                SousProbleme[elt+1, c] = SousProbleme[elt,c]
            end 
        end
    end
    sac = []
    indice = capacite+1
    i=n
    while i != 0
        if SousProbleme[i+1,indice] != SousProbleme[i,indice]
            sac= vcat(sac, [i])
            indice = indice - poid[i]
        end
        i=i-1
    end
    return(SousProbleme[n+1,capacite+1], sac)
end
                
                
```

```julia
function solveNdisplayKnap(filename)
    prix,poid, capacite = readKnaptxtInstance(filename)
    BestProfit, sac = sacADosDynamique(capacite,poid,prix)
    println(sac)
    println("\n******\n\nOptimal value = ", BestProfit)

end
```

#### Expliquer ce qu’est la programmation dynamique et la relation de récurrence associée à un problème de programmation dynamique


- La programation dynamique est une méthode pour résoudre les problèmes en résolvant des sous problèmes plus simple.
Elle permet de retrouver le résultat du problème que l'on cherche à l'aide des résultat précédement calculer et stocké. Pour chaque objet, pour une capacite donnée, la meilleure valeur est le maximum entre la solution sans l'objet et la solution avec l'objet : 
SousProbleme[i,j] = max(SousProbleme[i-1,j], SousProbleme[i-1, j-poid[i]] + prix[i])

#### Donner une courte argumentation de l’adéquation du résultat avec l’instance résolue


- Pour chaque valeur de elt, SousProbleme[i, j] prend la meilleur solution du sous-problème comportant les i-premiers objets et une capacite de sac de j. Dans ce cas, si l'on considère tous les objets et la capacité totale du sac à ods, nous tombons sur la cas général et la solution au sous-problème devient la solution du problème.

#### Expliquer le fonctionnement de votre algorithme


- On complète le sac objet par objet. pour un objet obj donné, on compare les valeurs du sac sans prendre l'objet (SousProbleme[obj, capacite+1]) et la solution en prenant l'objet qui correspond à la somme de la valeur de l'objet et la meilleure solution du sous-problème du sac-à-dos de capacité capacité-poid[obj].

#### En utilisant seulement des instances à au moins 20 objets, comparer les temps de calcul ob- tenus avec votre algorithme de programmation dynamique et votre branch-and-bound implé- menté lors du TP2 (ou à défaut la verson fournie avec la borne calculée via JuMP/relaxation linéaire). Observer les éventuelles différences. Comment expliquer ces observations ?



- En utilisant des instances d'au moins 20 objets nous pouvons remarquer que le temps de calcul nécessaire à la résolution du problème est nettement plus faible avec l'algorithme de programmation dynamique que celui du branch-and-bound. Cela est du au problème d'efficacité de la récursivité du programme branch-and-bound. En effet, pour N objets et un sac de capacité C, l'algorithme branch-and-bound va créer $2^N$ nœuds (moins grâce au test de sondages). Tandis que l'algorithme dynamique va créer un tableau de $N \times C $ cases. Cette différence est due à la sauvegarde des solutions des sous problèmes, le branch-and-bound ne réutilise pas les calculs déjà effectués. Ce problème d'efficacité fais par exemple écho au calcul de la suite de fibonacci ( $S_{n+1}\ =\ S_n\ +\ S_{n-1}$ ). En conclusion la programmation dynamique est plus rapide que le branch-and-bound car il évite le problème de la récursivité.
